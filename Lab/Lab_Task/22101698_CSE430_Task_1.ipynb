```python
import librosa
import librosa.display
import IPython.display as ipd
import numpy as np
import matplotlib.pyplot as plt
```

**Let's make the task interesting!** Record your own short audio clip using your PC microphone.
Choose a word or phrase that might sound similar when played backwards (for example, short symmetrical sounds like “wow”, “madam”, or “level”).

```python
audio_path = "/content/Recording.m4a"   # Load the audio file you recorded
signal, sr = librosa.load(audio_path, sr=None)

# Display basic information
print(" Audio loaded successfully!")
print(f"Sampling rate: {sr} Hz")
print(f"Total samples: {len(signal)}")
print(f"Duration: {len(signal)/sr:.2f} seconds")
ipd.display(ipd.Audio(signal, rate=sr)) # Play the audio
```

```python
signal_reversed = signal[::-1]      # Reverse the recorded audio
print("Reversed Audio (time reversed)")
ipd.display(ipd.Audio(signal_reversed, rate=sr)) # Play the audio
```

```python
plt.figure(figsize=(12, 8))

# Plot Original Audio
plt.subplot(2, 1, 1)
librosa.display.waveshow(signal, sr=sr, color='orange')
plt.title("Original Audio Waveform")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)

# Plot Reversed Audio
plt.subplot(2, 1, 2)
librosa.display.waveshow(signal_reversed, sr=sr, color='purple')
plt.title("Reversed Audio Waveform (Time Reversed)")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()
```
