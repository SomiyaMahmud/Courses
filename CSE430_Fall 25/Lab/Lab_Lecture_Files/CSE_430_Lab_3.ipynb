```python
import torch
import torch.nn as nn
```

```python
cnn_block = nn.Conv1d(in_channels=19, out_channels=64, kernel_size=3, stride=1, padding='valid',bias=True)

num_params = sum(p.numel() for p in cnn_block.parameters() )

print(f"Number of parameters in the CNN block: {num_params}")
```

```python
dummy_input = torch.randn(19,2048)
print(f"Input shape: {dummy_input.shape}")
output=cnn_block(dummy_input)
print(f"Output shape: {output.shape}")
```

## Multiple Blocks

```python

dummy_input = torch.randn(19,2048)
print(f"Input shape: {dummy_input.shape}")


cnn_block1 = nn.Conv1d(in_channels=19, out_channels=64, kernel_size=3, stride=1, padding='valid', bias = True)
cnn_block2 = nn.Conv1d(in_channels=64, out_channels=128, kernel_size=5, stride=1, padding='same', bias = True)
cnn_block3 = nn.Conv1d(in_channels=128, out_channels=256, kernel_size=7, stride=2, padding='valid', bias = True)


output1 = cnn_block1(dummy_input)
print(f"Output shape after block 1: {output1.shape}")

output2 = cnn_block2(output1)
print(f"Output shape after block 2: {output2.shape}")

output3 = cnn_block3(output2)
print(f"Output shape after block 3: {output3.shape}")
```

```python
total_params = sum(p.numel() for block in [cnn_block1, cnn_block2, cnn_block3] for p in block.parameters() if p.requires_grad)

print(f"Total number of parameters in the three blocks: {total_params}")
```

```python
print(output3)
```

```python
dummy_inp = torch.randn(64, 4, 4)

MaxPool_layer=nn.AvgPool2d(kernel_size=2)
output=MaxPool_layer(dummy_inp)
print(f"Output shape after MaxPool: {output.shape}")
```
