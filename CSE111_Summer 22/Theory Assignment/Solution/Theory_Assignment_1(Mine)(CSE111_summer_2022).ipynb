```python
#2
class ExamTracker:
    def __init__(self, course, title, exam_date=None):
        self.course = course
        self.title = title
        self.exam_date = exam_date
        self.modules = {}
        self.topic_list=[]
        print(f"New ExamTracker for {self.course} - {self.title} has been created...")
        self.count=0
        self.previous_module=0
    def add_topics(self,*topics,exam_date=None,module=None):
      present_module=module
      if self.count==0:
        self.previous_module=module
        self.count+=1
      if self.previous_module!=present_module :
         self.topic_list=[] 
         self.previous_module=module   

      if exam_date:  
        self.exam_date=exam_date
      
      if module:
        self.module=module
      else:
        self.module='default'
      if self.module not in self.modules:
        self.modules[module]={}
      for i in topics:
        self.topic_list.append(i)
        
      self.modules[self.module]={self.exam_date:self.topic_list}
      
    def update_date(self,new_date,module=None):
       self.topic_list=[]
       for i,j in self.modules.items():
        if i ==module :
          for k in j.values():
        
            self.topic_list.append(k)
       self.topic_list=self.topic_list[0] 
       dic={new_date:self.topic_list}

       self.modules[module]=dic
       print(f"{self.course} exam date for {self.title} - {module} " + \
              f"has been changed from {self.exam_date} to {new_date}")

    def get_details(self):
      
      output = "===================================\n"
      output += "Course: " + self.course + "\n"
      output += "Exam Title: " + self.title + "\n"
      output += "Modules:" + "\n"
      for i,j in self.modules.items():
        for x,k in j.items():
          output += "  " + self.title + " - " + str(i) + "\n"
          output += "  Exam Date: " + str(x) + "\n"
          for n in k:
            output += "    - " +str(n) + "\n"
          
      if len(self.modules) == 0:
            output += "  No modules have been created yet.\n"

      return output
    def get_topics(self,module=None):
      list1=[]
      if module==None:
        for i,j in self.modules.items():
          for x,y in j.items():
            for f in y:
              list1.append(f)
        return list1
      else:
        li=[]
        for i,j in self.modules.items():
          if i ==module :
            for k in j.values():
              li.append(k)
        return li[0]


    def print_details(self):
        print(self.get_details())




cse111_quiz = ExamTracker("CSE111", "Quiz")
cse111_quiz.add_topics("String Manipulations", module="1")
cse111_quiz.add_topics("List and Tuple", module="1")
cse111_quiz.add_topics("Dictionary", exam_date="15/06/2022",module="1")
cse111_quiz.add_topics("Functions", module="1")
cse111_quiz.update_date("20/06/2022", module="1")
cse111_quiz.add_topics("OOP Basics", exam_date="04/07/2022",module="2")
cse111_quiz.add_topics("Instance Variables and Functions","Method Overloading", module="2")
cse111_mid = ExamTracker("CSE111", "Midterm", "19/07/2022")
print(cse111_mid.get_details())
cse111_mid.add_topics(*cse111_quiz.get_topics())
cse111_mid.print_details()
cse111_quiz.add_topics("Operator Overloading","Encapsulation","Class Variable","Class & Static Method",module="3")
cse111_quiz.add_topics("Inheritance", "Method Overriding","Variable Overriding", "Abstract",exam_date="22/08/2022", module="4")
cse111_quiz.update_date("10/08/2022", module="3")
print(cse111_quiz.get_details())
cse111_final = ExamTracker("CSE111", "Final", "02/09/2022")
cse111_final.add_topics(*cse111_quiz.get_topics())
cse111_final.print_details()
print(cse111_quiz.get_topics())
print("================ GOOD LUCK ====================")
print(cse111_quiz.get_topics("3"))
```

```python
#3
class Department:
  def __init__(self,dept_initial,dept_name):
    self.dept_initial=dept_initial
    self.dept_name=dept_name
    self.course=[]
    

  def to_string(self):
    sum=0
    for i in self.course:
      sum+=i.credit
    output=f'Department of {self.dept_name} {[self.dept_initial]}\n'
    output+=f'Courses: {len(self.course)}\n Credits: {sum}\n'
    for j in self.course:
      output+=str(j)+'\n'
    return output
class Course:
  def __init__(self,code,title,credit,dept):
    self.code=code
    self.title=title
    self.credit=credit
    dept.course.append(self)

  def __str__(self):
    return f' - {self.code} [{self.credit}] | {self.title}'
    

cse = Department("CSE", "Computer Science and Engineering")
enh = Department("ENH", "English and Humanities")
mns = Department("MNS", "Mathematics and Natural Sciences")
cse110 = Course("CSE110", "Programming Language I", 3.0, cse)
cse111 = Course("CSE111", "Programming Language II", 3.0, cse)
eng091 = Course("ENG091", "Foundation Course", 0, enh)
eng101 = Course("ENG101", "English Fundamentals", 3.0, enh)
mat092 = Course("MAT092", "Remedial Course", 0.0, mns)
mat216 = Course("MAT216", "Linear Algebra and Fourier Analysis", 3, mns)
phy111 = Course("PHY111", "Principles of Physics I", 3, mns)
cse221 = Course("CsE112", "Algorithms", 0, mns)
print(cse.to_string())
print(enh.to_string())
print(mns.to_string())
cse221.update(code="CSE221", credit=3, department=cse)
print("\n", "=" * 25, sep="\n")
print("||" + " " * 5 + "Summer 2022" + " " * 5 + "||")
print("=" * 25, "\n", sep="\n")
print(cse.to_string())
print(mns.to_string())
print(cse221 == cse.get_course_by_code("CSE221"))
print(cse.get_course_by_code("CSE110").to_string())
```
