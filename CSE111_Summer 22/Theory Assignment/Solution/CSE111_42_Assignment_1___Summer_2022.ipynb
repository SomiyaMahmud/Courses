```python
# Question 2

class ExamTracker:
    def __init__(self, course, title, exam_date="Not Defined", *topics):
        self.course = course
        self.title = title
        self.exam_date = exam_date
        self.modules = {}
        print(f"New ExamTracker for {course} - {title} has been created...")

    def update_date(self, exam_date, module=None):
        if not module:
            module = "default"
        if module not in self.modules or len(self.modules[module]) < 1:
            self.modules[module] = [exam_date]
        old_date = self.modules[module][0]
        self.modules[module][0] = exam_date
        print(f"{self.course} exam date for {self.title} - {module} has been changed from {old_date} to {exam_date}")

    def add_topics(self, *topics, exam_date=None, module=None):
        if not module:
            module = "default"
        if module not in self.modules:
            self.modules[module] = [None]
        if not exam_date:
            if self.modules[module][0] == None:
                exam_date = self.exam_date
            else:
                exam_date = self.modules[module][0]
        if self.modules[module][0] == None or self.modules[module][0] == "Not Defined":
            self.modules[module][0] = exam_date
        elif self.modules[module][0] != exam_date:
            self.update_date(exam_date, module)

        for topic in topics:
            self.modules[module].append(topic)
        
    def get_topics(self, module=None):
        topics = []
        if module:
            for topic in self.modules[module][1:]:
                topics.append(topic)
        else:
            for module in self.modules:
                for topic in self.modules[module][1:]:
                    topics.append(topic)

        return topics
            

    def get_details(self):
        output = "===================================\n"
        output += "Course: " + self.course + "\n"
        output += "Exam Title: " + self.title + "\n"
        output += "Modules:" + "\n"
        for module, topics in self.modules.items():
            output += "  " + self.title + " - " + module + "\n"
            output += "  Exam Date: " + topics[0] + "\n"
            for topic in topics[1:]:
                output += "    - " + topic + "\n"
        if len(self.modules) == 0:
            output += "  No modules have been created yet.\n"
        return output

    def print_details(self):
        print(self.get_details())


cse111_quiz = ExamTracker("CSE111", "Quiz")
cse111_quiz.add_topics("String Manipulations", module="1")
cse111_quiz.add_topics("List and Tuple", module="1")
cse111_quiz.add_topics("Dictionary", exam_date="15/06/2022", 
                       module="1")
cse111_quiz.add_topics("Functions", module="1")
cse111_quiz.update_date("20/06/2022", module="1")
cse111_quiz.add_topics("OOP Basics", exam_date="04/07/2022", 
                       module="2")
cse111_quiz.add_topics("Instance Variables and Functions", 
                       "Method Overloading", module="2")
cse111_mid = ExamTracker("CSE111", "Midterm", "19/07/2022")
print(cse111_mid.get_details())
cse111_mid.add_topics(*cse111_quiz.get_topics())
cse111_mid.print_details()
cse111_quiz.add_topics("Operator Overloading",
                       "Encapsulation", 
                       "Class Variable", 
                       "Class & Static Method", 
                       module="3")
cse111_quiz.add_topics("Inheritance", "Method Overriding",
                       "Variable Overriding", "Abstract",
                       exam_date="22/08/2022", module="4")
cse111_quiz.update_date("10/08/2022", module="3")
print(cse111_quiz.get_details())
cse111_final = ExamTracker("CSE111", "Final", "02/09/2022")
cse111_final.add_topics(*cse111_quiz.get_topics())
cse111_final.print_details()
print(cse111_quiz.get_topics())
print("================ GOOD LUCK ====================")
print(cse111_quiz.get_topics("3"))
```

```python
# Question 3

class Course:
    def __init__(self, code, title, credit, department):
        self.code = code
        self.title = title
        self.credit = float(credit)
        self.department = department
        department.register_courses(self)

    def update(self, **kwargs):
        for k, v in kwargs.items():
            if k in ["code", "title", "credit", "department"]:    
                if k == "department":
                    self.department.remove_courses(self)
                    v.register_courses(self)
                setattr(self, k, v)
        self.credit = float(self.credit)

    def to_string(self):
        return f"{self.code} [{self.credit}] | {self.title}"
    
class Department:
    def __init__(self, abbreviation, title):
        self.abbreviation = abbreviation
        self.title = title
        self.courses = []
    
    def get_course_by_code(self, code):
        for course in self.courses:
            if course.code == code:
                return course
        return None

    def register_courses(self, *courses):
        for course in courses:
            self.courses.append(course)
    
    def remove_courses(self, *courses):
        for course in courses:
            index = -1
            for i in range(len(self.courses)):
                if self.courses[i] == course:
                    index = i
            if index >= 0:
                del self.courses[index]

    def to_string(self):
        result = f"Department of {self.title} [{self.abbreviation}]\n" + \
                 f"Courses: {len(self.courses)}\n"
        total_credits = 0
        courses = ""

        for course in self.courses:
            courses += f"  - {course.to_string()}\n"
            total_credits += course.credit

        result = result + \
                 f"Credits: {total_credits}\n" + \
                 f"{courses}" + \
                 f"===================================================="
        
        return result

cse = Department("CSE", "Computer Science and Engineering")
enh = Department("ENH", "English and Humanities")
mns = Department("MNS", "Mathematics and Natural Sciences")
cse110 = Course("CSE110", "Programming Language I", 3.0, cse)
cse111 = Course("CSE111", "Programming Language II", 3.0, cse)
eng091 = Course("ENG091", "Foundation Course", 0, enh)
eng101 = Course("ENG101", "English Fundamentals", 3.0, enh)
mat092 = Course("MAT092", "Remedial Course", 0.0, mns)
mat216 = Course("MAT216", "Linear Algebra and Fourier Analysis", 3, mns)
phy111 = Course("PHY111", "Principles of Physics I", 3, mns)
cse221 = Course("CsE112", "Algorithms", 0, mns)

print(cse.to_string())
print(enh.to_string())
print(mns.to_string())

cse221.update(code="CSE221", credit=3, department=cse)

print("\n", "=" * 25, sep="\n")
print("||" + " " * 5 + "Summer 2022" + " " * 5 + "||")
print("=" * 25, "\n", sep="\n")
print(cse.to_string())
print(mns.to_string())
print(cse221 == cse.get_course_by_code("CSE221"))
print(cse.get_course_by_code("CSE110").to_string())
```
