```python
class MetaClass(type):
    __instances = {}

    @classmethod
    def __getitem__(cls, k):
        return cls.__instances[k]

    @classmethod
    def __setitem__(cls, k, v):
        cls.__instances[k] = v

    @classmethod
    def __len__(cls):
        return len(cls.__instances)

    @classmethod
    def __delitem__(cls, key):
        cls.__instances.__delitems__(key)
    
    @classmethod
    def __iter__(cls):
        return iter(cls.__instances)

class Dell(metaclass=MetaClass):
    __personal = {}
    def __init__(self, region):
        self.region = region
        Dell[region] = self

    def __str__(self):
        return self.region

dell_bd = Dell("Bangladesh")

print(Dell["Bangladesh"])
```
