```python
# Set B
class Employee:
    employees = []
    total_salary = 0.0
    def __init__(self, name, working_hours):
        self.__id = len(self.employees) + 1
        self.__name = name
        self.__salary = 0.0
        self.__working_hours = 0
        self.setWorkingHours(working_hours)
        self.employees.append(self)

    def getName(self):
        return self.__name

    def setName(self, name):
        self.__name = name

    def setWorkingHours(self, working_hours):
        Employee.total_salary -= self.__salary
        self.__working_hours = working_hours
        self.__salary = self.calculate_salary(working_hours)
        Employee.total_salary += self.__salary
    
    def getString(self):
        return f"< ID: {self.__id}, Name: {self.__name}, WH: {self.__working_hours} hours, Salary: ${self.__salary:.2f} >"

    @classmethod
    def flush(cls):
        print("Total employees:", len(cls.employees))
        print(f"Total salary: ${cls.total_salary}")
        for employee in cls.employees:
            print(employee.getString())
            print(f"<<< Employee receives {employee.__salary * 100 / cls.total_salary:.2f}% salary of total salary >>>")

    @staticmethod
    def calculate_salary(working_hours):
        if working_hours < 5:
            return working_hours * 400
        elif working_hours < 7:
            return working_hours * 350
        return working_hours * 300


Employee.flush()
print('########################')
robert = Employee("RoBERTa", 8)
print(robert.getString())
print('------------------------')
Employee.flush()
print("========================")
james = Employee("James", 3)
albert = Employee("Albert", 12)
robert.setName("Robert")
robert.setWorkingHours(10)
Employee.flush()
print("========================")
print(Employee.calculate_salary(15))
```

```python
# Set C
class Country:
    countries = []
    total_countries = 0
    total_asset = 0
    def __init__(self, name):
        self.__name = name
        self.__population = 0
        self.__asset = 0.0
        self.__id = self.total_countries
        self.countries.append(self)
        Country.total_countries += 1

    def setPopulation(self, population):
        Country.total_asset -= self.__asset
        self.__population = population
        if population < 100:
            self.__asset = population * 300.0
        elif population < 150:
            self.__asset = population * 400.0
        else:
            self.__asset = population * 350.0
        Country.total_asset += self.__asset

    @classmethod
    def details(cls):
        print("Total number of country(ies):", cls.total_countries)
        print(f"Total assets: ${cls.total_asset}")
        for country in cls.countries:
            country.info()
            print(f"((( Country consists of total asset's: {round(country.__asset * 100 / cls.total_asset, 2)}% )))")

    def info(self):
        print(f"( ID: {self.__id}, Name: {self.__name}, Population: {self.__population}M, Asset: ${self.__asset} )")

Country.details()
print('########################')
bangladesh = Country("Bangladesh")
bangladesh.setPopulation(64)
bangladesh.info()
print('------------------------')
Country.details()

print('========================')

china = Country("China")
japan = Country("Japan")
japan.setPopulation(125)
china.setPopulation(1402)
bangladesh.setPopulation(164)
Country.details()
```
