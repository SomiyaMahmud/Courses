```python
#sir
class ExamTracker:
    def __init__(self, course, title, exam_date=None):
        self.course = course
        self.title = title
        self.exam_date = exam_date
        self.modules = {}
        print(f"New ExamTracker for {course} - {title} has been created...")

    def update_date(self, exam_date, module=None):
        if not module:
            module = "default"
        if module not in self.modules or len(self.modules[module]) < 1:
            self.modules[module] = [exam_date]
        old_date = self.modules[module][0]
        self.modules[module][0] = exam_date
        print(f"{self.course} exam date for {self.title} - {module} " + \
              f"has been changed from {old_date} to {exam_date}")

    def add_topics(self, *topics, exam_date=None, module=None):
        if not module:
            module = "default"
        if module not in self.modules:
            self.modules[module] = [None]
        if exam_date == None and self.modules[module][0] == None:
            exam_date = self.exam_date
        if self.modules[module][0] == None:
            self.modules[module][0] = exam_date
        elif exam_date != None and self.modules[module][0] != exam_date:
            self.update_date(exam_date, module)

        for topic in topics:
            self.modules[module].append(topic)            

    def get_details(self):
        output = "===================================\n"
        output += "Course: " + self.course + "\n"
        output += "Exam Title: " + self.title + "\n"
        output += "Modules:" + "\n"

        for module, topics in self.modules.items():
            exam_date = topics[0]
            output += "  " + self.title + " - " + module + "\n"
            if exam_date == None:
                exam_date = "Not Defined"
            output += "  Exam Date: " + exam_date + "\n"
            for topic in topics[1:]:
                output += "    - " + topic + "\n"

        if len(self.modules) == 0:
            output += "  No modules have been created yet.\n"

        return output

    def print_details(self):
        print(self.get_details())

cse111_quiz = ExamTracker("CSE111", "Quiz")
cse111_quiz.add_topics("String Manipulations", "List and Tuple", module="1")
cse111_quiz.add_topics("Dictionary", "Functions", exam_date="15/06/2022", module="1")
cse111_quiz.update_date("20/06/2022", module="1")
cse111_quiz.print_details()
cse111_mid = ExamTracker("CSE111", "Midterm", "19/07/2022")
print(cse111_mid.get_details())
cse111_mid.add_topics("Up to Quiz 2")
cse111_mid.print_details()
```
